### Q1.A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing

```python
import pandas as pd

df = pd.DataFrame({
    'Student_ID': ['S101', 'S102', 'S103', 'S104', 'S105', 'S106',
                   'S107', 'S108', 'S102', 'S109', 'S110', 'S111',
                   'S112', 'S113', 'S114'],
    'Student_Name': ['Asha', 'Ravi', 'Neha', 'Karan', 'Meera', 'Arjun',
                     'Priya', 'Rahul', 'Ravi', 'Sneha', 'Varun', 'Diya',
                     'Aman', 'Nisha', 'Kabir'],
    'Department': ['IT', 'it', 'I.T.', 'HR', 'HR', 'Finance',
                   'finance', 'IT', 'it', 'I.T.', 'HR', 'Finance',
                   'IT', 'HR', 'Finance'],
    'Assessment': ['Python'] * 15,
    'Marks': [85, 78, 92, None, 'Absent', 0, 67, 75, 78,
              'NA', '88 marks', 91, None, 64, 105]
})

print(df)
```

## Using Pandas, perform the following cleaning operations in sequence:

### 1.Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().


```python
print(df.head())
print(df.info())
print(df.isnull().sum())
```

### 2.Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

```python
df['Department'] = df['Department'].replace({
    'it': 'IT',
    'I.T.': 'IT',
    'finance': 'Finance'
})

print(df['Department'])
```

**Explanation:** The `replace()` method standardizes inconsistent department names so that `IT`, `it`, and `I.T.` become `IT`, while `Finance` and `finance` become `Finance`.

duplicates = df[df.duplicated(keep=False)]

print(duplicates)

df = df.drop_duplicates()

print(df)

**Explanation:** `duplicated()` identifies repeated records, while `drop_duplicates()` removes exact duplicate rows and keeps the first occurrence by default.

### 3.Identify and remove exact duplicate student assessment records.


```python
df['Marks'] = df['Marks'].replace({
    'Absent': pd.NA,
    'NA': pd.NA
})

df['Marks'] = (
    df['Marks']
    .astype('string')
    .str.extract(r'(\d+(?:\.\d+)?)', expand=False)
)

df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')

print(df[['Student_ID', 'Student_Name', 'Marks']])
```

**Explanation:** The `replace()` method treats `Absent` and `NA` as missing values. The regular expression extracts numeric values from entries such as `88 marks`, and `pd.to_numeric()` converts the cleaned values into a numeric format. Invalid or missing values become `NaN`.

### 4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.

```python
df['Marks_Status'] = df['Student_ID'].map({
    'S104': 'Not Entered',
    'S105': 'Absent',
    'S109': 'Needs Investigation',
    'S112': 'Not Entered'
}).fillna('Recorded')

print(df[['Student_ID', 'Student_Name', 'Marks', 'Marks_Status']])
```

### 5.Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.


```python
invalid_mask = df['Marks'].notna() & ~df['Marks'].between(0, 100)

invalid_marks = df.loc[invalid_mask]

print(invalid_marks[['Student_ID', 'Student_Name', 'Marks']])
```

**Explanation:**  Marks outside the valid 0–100 range are treated as invalid and should be investigated before including them in analysis.

### 6.Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.
